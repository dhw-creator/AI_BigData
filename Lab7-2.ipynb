```python
casts.groupby([decade,'type']).size()
```

```python
c_dec2 = casts.groupby([decade, 'type']).size()
```

```python
c_dec2.plot(kind='bar')
plt.show()
```

```python
c_dec3 = casts.groupby(['type',decade]).size()
c_dec3.plot(kind = 'bar')
plt.show()
```

```python
c_dec3
```

```python
c_dec3.unstack() #multiindex -> column
```

```python
un_c_dec3 = c_dec3.unstack()
un_c_dec3.plot(kind='bar')
plt.show()
```

```python
un_c_dec3.T
```

```python
un_c_dec3.T.plot(kind='bar')
plt.show()
```
