```python
import numpy as np
```

```python
def f(t):
    return np.exp(-t)*np.cos(2*np.pl*t)
```

```python
t1=np.arange(0,5,0.1)
t2=np.arange(0,5,0.02)
```

```python
len(t1)
```

```python
len(t2)
```

```python
plt.figure(figsize=(6,4),dpi = 150)
plt.subplot(211)
plt.plot(t1,f(t1),'gd',t2,f(t2),'r.')
plt.title('(a) 첫번째 그래프')
plt.subplot(2,1,2)
plt.plot(t2,np.cos(2*np,pi*t2),'m-.')
plt.title('(b) 두번째 그래프')
plt.suptitle('전체 그래프 제목')
plt.show()
```
